# PSELDNets — outdoor_siren_v5 学習（train 60 / val 20、マルチクラス拡張版、run2=200epoch）

**マルチクラスSELD拡張（v3と同じ土俵、対象クラスを4つに増やした版）**: v3と同一の速度・距離・
SNR・SIR・発音区間レンジだが、検出対象を Siren 単独から **Siren / Horn(クラクション) /
BackupBeep(バック警告音) / BikeBell(自転車ベル) の4クラス**に拡張（各20本ずつ均等）。
妨害音（車の走行音）も実録音ではなく**クリーン合成**に統一（対象と妨害音の音質を揃え、
音質の違いがショートカット手がかりになるのを防ぐため）。

**run2の変更点**: run1(70epoch)はepoch70時点でもSELD_scrが下降し続けたまま終了＝学習不足
だったため、**epoch数を200に増やして再学習**する（4クラス分割で1クラスあたりの学習データが
v3の1/4になり、v3で妥当だった70epochでは足りなかったと推定）。

クラス辞書は屋外10クラス辞書ではなく、本データセット専用の4クラス辞書
（`configs/cls_indices_v5.tsv`）を使う。モデルの出力ヘッド(`tscam_conv`)は
クラス数×9chで自動的に再構築される（事前学習ckptのヘッドは毎回スキップされ
自動再初期化されるため、クラス数が変わっても問題なくロードできる）。

## 11. 結果の確認

`val/macro` の ER / F / LE / LR / **SELD_scr** を全エポック分抽出する。

**v5 run2の見どころ**: run1(70epoch)の最終値 ER 0.396/F 65.6%/LE 18.8°/SELD_scr 0.219 から

疑う次段階に進む。ここで出る指標は**4クラス全体の平均**なので、クラスごとの内訳（Sirenは
得意でBikeBellは苦手、等の差）を見るには別途クラス別の分析
（`scripts/step7_error_anatomy.py`のマルチクラス対応が必要、現状は未実装）。

- データ生成条件の全記録はローカルの `outdoor_seld_e2e/out/dataset_outdoor_siren_v5/`
  （`inspection.csv`=検品結果, `work/*/scene.json`=各クリップの条件、`hazard_class`欄に

- SNR/SIR はサイレン等の発音区間の W チャンネル基準で定義
- 結果を報告するときは ER / F / LE / LR / SELD_scr を併記する
- **experiment_nameの使い回しに注意**（v3のrun1 NaN崩壊事故参照。データを変えたら
  必ず新しいexperiment_nameにする）
- クラス別の内訳分析（どのクラスが苦手か等）は、この結果だけでは分からない。
